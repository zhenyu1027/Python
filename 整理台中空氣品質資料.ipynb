```python
import json
from io import TextIOWrapper
from zipfile import ZipFile
from io import BytesIO
import requests
import urllib
from urllib.request import urlretrieve
import ssl

def fullwidth(s):             # 自訂一個 Function 轉換全形半形
    result = ""
    for c in str(s):          # 讓迴圈一一讀取，str()是讓放入得值先變成字元型態
        if c == ')':          # 此檔案因為碰到 ')'時，只佔一格， '(' 則佔兩格，所以導致印出來還是歪歪的，所以做特殊處理，讓碰到 ')' 再往後一格
           result += ') '
        else:                 
           if 33 <= ord(c) <= 126:    # '!' ~ '~' ASCII CODE 33~126 ， 利用 ord() 來得到這些字元的 ASCII code 進行比較，如果不再這個範圍內則不須改成全形
              result += chr(ord(c) + 65248)       # 將這個範圍的字元都改成全形
           else:
              result += c     # 已經是全形的只要一一讀取，存回 result 即可
    return result             # 回傳result


#以下 24行~31行 為下載與讀取一個在網路上其它主機內已經被 ZIP 壓縮過的檔案
ssl._create_default_https_context = ssl._create_unverified_context

url = "https://opendata.taichung.gov.tw/api/v1/dataset.all.resource.download?pid=b94015c6-14f3-47c1-8870-719b26bb6093"    
#檔案的網址，因為此檔案會更新不是靜態的，因此需要動態調整，讓每次執行都是下載並讀取最新的檔案

urlretrieve(url, "data/jsondata.zip")

zip_file = ZipFile("data/jsondata.zip")

#print(zip_file.namelist(), '\n')    # 此行為自己測試用

for name in zip_file.namelist():              # 找尋zip檔中所有的檔名，一一讀取
     if name.lower().endswith('.json'):       # 這是在處理檔案大小寫問題，僅限json檔
        with zip_file.open(name, 'r') as jsonfile:   # 直接打開zip檔，讀取用 'r'，取名叫 jsonfile
             data = json.load(TextIOWrapper(jsonfile, 'utf-8-sig'))   
             # json.load() 把 JSON 變成 Python 資料， TextIOWrapper() 把 binary 轉成文字'utf-8-sig'處理 BOM（避免亂碼）
             columns = list(data[0].keys())   # 將第一筆資料( data[0] )的欄位名稱( .keys() )轉換成 list，存到 columns
            
             print("『 " + name + " 』總共有 " + str(len(columns)) + " 個欄位，分別為：")   #印出總共多少欄位，分別有哪些欄位名稱
             # 另一種寫法 print(f"『 {name} 』總共有 {len(columns)} 個欄位，分別為：")
             
             #print(len(columns))    # 此行為測試用
             i = 0                   # 類似一個計數器
             for col in columns:     # 用迴圈一一讀取 columns(就是欄位名稱)
                 if i != len(columns)-1:     # 因為要將所有欄位加上『 』，且中間要用逗點隔開，但最後一個欄位名稱不需要逗點，因此如果非最後一個欄位就要輸出逗點
                    print("『 " + col +" 』," , end='')   
                 else:
                    print("『 " + col +" 』" )    # 最後一個欄位不須輸出逗點  
                 i += 1              # 此計數器可以表示讀到第幾個欄位名稱
             print()
            
             fields = ['Id', 'Town', 'Landmark', 'Temp', 'PM2.5', 'Hum']     # 只需列出這幾個欄位的資料
            
             maxlen = {}        # 因為要整齊輸出資料，所以要先知道每個欄位最多佔幾格，先建立一個空的 dict，要用來記錄每個欄位最大長度
             for f in fields:   # 先初始化讓每個長度都為 0
                 maxlen[f] = 0
             for item in data:  # 一一讀取每一筆資料
                 #print(item)   # 此行為測試用
                 for f in fields: # 一一讀取每個欄位
                     #print(f)  # 此行為測試用
                     #print(item[f])   # 此行為測試用
                     val = fullwidth(item[f])   # 將 item[f] 讀取出來的值放入自訂的 fullwidth() 裡做半形轉全形的處理
                     maxlen[f] = max(maxlen[f], len(str(item[f])))  # 比較，留下大的表示照樣才能抓到最大長度
             #print(maxlen)     # 此行為測試用

             for h in fields:   # 一一讀取每個欄位
                 space = maxlen[h] - len(fullwidth(h))  # 計算需要補幾個空格
                 text = h + "　" * (space + 2)          # 補齊欄位長度，並讓欄位不要連在一起
                 print(fullwidth(text), end='')         # 因為剛剛都是半形，所以要再經過 fullwidth() 轉為全形
             print()
             print()
            
             for item in data:                          # 一一讀取每筆資料      
                 for f in fields:                       # 一一讀取每個欄位，按照順序取值
                     value = str(item[f])               # 先將每筆資料 ( item[f] ) 用str()轉成字串，因為string 無法與 integer 利用 + 相連 
                     space = maxlen[f] - len(value)     #該欄最大長度 - 目前這筆資料長度，找出需要補幾個空格
                     text = value + "　" * (space + 2)  #補齊欄位長度，並讓欄位不要連在一起
                     print(fullwidth(text), end='')     #利用fullwidth()轉成全形
                 print()    
                
             jsonfile.close()
```
