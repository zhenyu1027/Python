```python
import json
from io import TextIOWrapper
from io import BytesIO
import requests
import urllib
from urllib.request import urlretrieve
import ssl
```

```python
def fullwidth(s):
    result = ""
    for c in str(s):
        if c == ")":  # 也可以寫成 chr(41) # 如果遇到 *)* 要改成 全形 *）*  (*為辨識半形 全形，以免自己忘記) 
            result = result + "）" # 也可以寫成 chr(65289)
        else:
            if 33 <= ord(c) <= 126: # ord() 用來取得ASCII code，33~126 是可見半形字字元
                result += chr(ord(c) + 65248) # 加 65248 就變成全形字元
            else:
                result += c
    return result
            
ssl._create_default_https_context = ssl._create_unverified_context
# ssl._create_unverified_context 是一個「不驗證憑證」的設定 
# 指定給 ssl._create_default_https_context，等於告訴 Python：以後建立 HTTPS 連線時，預設都用「不驗證」的方式

url = "https://tcgbusfs.blob.core.windows.net/dotapp/youbike/v2/youbike_immediate.json"

path = "data/台北市Youbike2.0jsondata.json" 

urlretrieve(url, path) # 下載網址內容，存到 data 資料夾

with open(path,"r" , encoding = "utf-8-sig") as file:
    data = json.load(file)
    
print(data[0],"\n")
print(list(data[0].items()),"\n") # 取得所有 key & value
print(list(data[0].keys()),"\n") # 取得所有 key
print(list(data[0].values()),"\n") #取得所有 value
print(len(data[0])) # 18個欄位
print(type(data)) # data 是屬於 list
print(type(data[0])) # list中的 dic
```
