```python
import pandas as pd
import json
from urllib.request import urlretrieve
```

```python
url = "https://tcgbusfs.blob.core.windows.net/dotapp/youbike/v2/youbike_immediate.json"
path = "data/台北市Youbike2.0jsondata.json" 
urlretrieve(url, path)
with open(path,"r" , encoding = "utf-8-sig") as file:
    data = json.load(file)
```

```python
df = pd.DataFrame(data)
df.head()
```

```python
df.columns
```

```python
df["sna"]
```

```python
df = df.rename(columns={"sna": "場站", "sarea": "場站區域", "ar": "地點", "available_rent_bikes": "目前車輛數", "mday": "更新時間"})
df
```

```python
result = df[["場站", "場站區域", "地點", "目前車輛數", "更新時間"]]
result.to_excel("data/YouBike2.0整理後資料.xlsx", index=False)
```
