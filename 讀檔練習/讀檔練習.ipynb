```python
with open("test2.txt", mode="w", encoding="utf-8") as file:
    file.write("這個寫法不用close\n5\n3")
```

```python
sum = 0
with open("test2.txt", mode="r", encoding="utf-8") as file:
    for line in file:
        line = line.strip()
        if(line.isdigit()):
            sum+=int(line)
print(sum)
```

# 使用 JSON 格式讀取 寫檔

```python
import json
```

```python
with open("test.json", mode="w") as file:
    json.dump({"name": "James", "version": "1.2.5"}, file)
```

```python
with open("test.json", mode="r") as file:
    data = json.load(file)
print(data)
print("name:", data["name"])
print("version:", data["version"])
```

```python
data["name"] = "Daisy"
with open("test.json", mode="w") as f:
    json.dump(data, f)
print(data)
print("name:", data["name"])
print("version:", data["version"])
```
